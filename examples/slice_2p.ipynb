```python
import tifffile
import os
```

```python
# Load OME-TIFF file
folderpath = r'Z:\TM_Lab\Edward\Cerebellum_slice_2p\20250430_fem_dob0311_AER3'
filename = 'slice1_acq1'
with tifffile.TiffFile(os.path.join(folderpath,filename + '.ome.tiff')) as tif:
    # tif.series[0] contains all the image data; squeeze to remove singleton dimensions
    image = tif.series[0].asarray()

```

```python
image.shape
```

```python
with tifffile.TiffWriter(os.path.join(folderpath, filename+'.tif'), bigtiff=True) as tif_writer:
    for i in range(image.shape[1]-2): # check bad end frames
        tif_writer.write(image[0,i,:,:].squeeze())
```
